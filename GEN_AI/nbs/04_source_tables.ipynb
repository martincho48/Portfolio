```python
# Zjistíme tvůj aktuální katalog a schéma v Unity Catalogu
current_catalog = spark.sql("SELECT current_catalog()").collect()[0][0]
current_schema = spark.sql("SELECT current_schema()").collect()[0][0]

UC_PREFIX = f"{current_catalog}.{current_schema}"
print(f"Tvůj agent bude číst tabulky z: {UC_PREFIX}")
```

```python
import pandas as pd

dummy_table = pd.DataFrame({
    "product": ["A", "B", "C"],
    "price": [100, 200, 300],
    "quantity": [1, 4, 2]
})


# 1. Převedeme Pandas DataFrame na Spark DataFrame
spark_df = spark.createDataFrame(dummy_table)

# 2. Uložíme jako Delta tabulku (přepíše se, pokud už existuje)
spark_df.write.format("delta").mode("overwrite").saveAsTable(f"{UC_PREFIX}.delta_table")
```
